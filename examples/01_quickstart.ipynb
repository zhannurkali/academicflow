# academicflow quickstart

Every feature through the plain Python API. For the notebook magics, see `02_full_paper_workflow.ipynb`.

```python
import academicflow as af

text = "Students in data science spend a lot of time in Jupyter notebooks, but they still write their papers in separate tools. NLP tools for grammar checking are widely available, but they don't understand the conventions of academic writing. Natural language processing (NLP) has changed how we analyse text [@lecun2015], and large language models (LLM) can now write fluent paragraphs [@vaswani2071]. However, it's obviously not enough to get fluent text; a paper must also follow the style of the the target venue. Basically, students need a tool that checks their stuff where they already work. In this work we collected a data set of twenty student drafts and built a really simple pipeline around it."
af.__version__
```

## Text statistics

```python
af.analyze(text)
```

## Style check

```python
for issue in af.check(text, profile='apa', section='introduction'):
    print(f'{issue.severity:8} {issue.rule:20} {issue.message}')
```

## Citations

```python
bib = af.Bibliography.from_file('refs.bib')
print(bib.keys())
rendered, references = af.render(text, bib, style='ieee')
print(rendered)
print(*references, sep='\n')
```

```python
af.check_integrity(text, bib)
```

## Summary and abstract

```python
af.summarize(text, max_words=40)
```
